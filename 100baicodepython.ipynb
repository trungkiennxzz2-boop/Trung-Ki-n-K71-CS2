# 0006 - Tính chu vi và diện tích của một tam giác
Đề bài
Viết chương trình tính chu vi (cv) và diện tích (s) một tam giác có 3 cạnh a, b, c là các
số nguyên theo công thức cv=a+b+c, s=sqrt(p*(p-a)*(p-b)*(p-c)) với p=cv/2 (3 số a, b, c
bảo đảm là 3 cạnh của 1 tam giác).
Dữ liệu vào
Gồm 1 dòng có 3 số nguyên cách nhau bởi dấu cách theo thứ tự là a, b và c.
Dữ liệu ra
Là chu vi và diện tích nằm trên cùng một dòng, cách nhau bởi dấu cách, chu vi làm tròn
đến 1 chữ số thập phân, diện tích làm tròn đến 3 chữ số thập phân.
Ví dụ
Input #1
2 3 4
Output #1
9.0 2.905
Input #2
3 4 5
Output #2
12.0 6.000

```python
a ,b ,c = map(int,input().split())
import math
if (a+b)>c and (a+c)>b and (b+c)>a :
    cv = a+b+c
    p= cv/2
    dt=math.sqrt(p*(p-a)*(p-b)*(p-c))
    print(f"{cv:.1f} {dt:.3f}")
else:
    print("không là tam giác")
```

các số trong mảng
Dữ liệu vào
Dòng 1 là số nguyên dương n, dòng 2 là n số nguyên.
Dữ liệu ra
Trung bình cộng các số trong mảng làm tròn đến 1 chữ số thập phân
Ví dụ
Input #1
14

Output #1
7.3

```python

def tongm(a):
    k=0
    for i in a:
        k+=i
    return k
m=int(input())
n=list(map(int,input().split()))
if len(n)==m:
    tbc=tongm(n)/len(n)
    print(f"{tbc:.1f}")
    
```

# 0096 - Viết chương trình con tính tổng các phần tử chẵn
trong 1 mảng
Đề bài
Viết chương trình con tính tổng các phần tử chẵn trong 1 mảng, áp dụng tính trung bình
cộng các số chẵn trong mảng
Dữ liệu vào
Dòng 1 là số nguyên dương n, dòng 2 là n số nguyên.
Dữ liệu ra
Trung bình cộng các số chẵn trong mảng, làm tròn đến 2 chữ số thập phân
Ví dụ
Input #1
14
16 28 -16 -4 -13 11 28 8 10 21 4 30 -7 -14 
Output #1
9.00

```python
def chan(x):
    tong=0
    for i in x:
        if i%2==0:
            tong+=i
    return tong
m = int(input("Nhập số lượng phần tử n: "))
n = [int(x) for x in input(f"Nhập {n} số cách nhau bằng khoảng trắng: ").split()][:n]

print(f"{chan(n)/m:.2f}")
```
